## Introduction to Reinforcement Learning Using Python

### Aim
To understand the basic concepts of Reinforcement Learning by simulating a simple agent-environment interaction using Python.

### Problem Statement
A robot is standing at the entrance of a maze. It can choose between different actions. Only the Forward action leads toward the goal and provides a positive reward.

### Learning Objectives
1. Understand Agent, Environment, State, Action and Reward.
2. Simulate agent-environment interaction.
3. Understand how rewards influence an agent's decisions.
4. Implement a simple reward-based decision-making system.
5. Calculate rewards obtained over multiple trials.

```python
import random

# Set seed for reproducible results
random.seed(42)

print("Reinforcement Learning Simulation")
print("Libraries imported successfully.")
```

```python
# Basic components of Reinforcement Learning

agent = "Robot"
environment = "Maze"
state = "Starting Position"

actions = ["Move Left", "Move Right", "Move Forward"]

print("Agent:", agent)
print("Environment:", environment)
print("Initial State:", state)
print("Available Actions:", actions)

print("\nExplanation:")
print("Agent: The robot that makes decisions.")
print("Environment: The maze in which the robot operates.")
print("State: The current position of the robot.")
print("Action: The movement selected by the robot.")
print("Reward: Feedback received after performing an action.")
```

```python
# Define the environment

def maze_environment(action):

    if action == "Move Forward":
        reward = 10
        result = "Goal Reached!"

    elif action == "Move Left":
        reward = -2
        result = "Wrong Direction!"

    elif action == "Move Right":
        reward = -2
        result = "Wrong Direction!"

    else:
        reward = 0
        result = "Invalid Action!"

    return reward, result
```

```python
# Define available actions

actions = ["Move Left", "Move Right", "Move Forward"]

# Agent selects an action randomly
selected_action = random.choice(actions)

# Environment responds to the action
reward, result = maze_environment(selected_action)

# Display results
print("Agent: Robot")
print("Selected Action:", selected_action)
print("Environment Response:", result)
print("Reward Obtained:", reward)
```

```python
# Simulate 10 interactions

total_reward = 0

print("Robot Maze Simulation")
print("-" * 40)

for trial in range(1, 11):

    action = random.choice(actions)

    reward, result = maze_environment(action)

    total_reward += reward

    print("Trial:", trial)
    print("Action:", action)
    print("Result:", result)
    print("Reward:", reward)
    print("-" * 40)

print("Total Reward:", total_reward)
```

```python
# Define five available actions

actions = [
    "Move Left",
    "Move Right",
    "Move Forward",
    "Move Backward",
    "Stay"
]

print("Available Actions:")

for action in actions:
    print("-", action)
```

```python
# Define reward values for each action

reward_values = {
    "Move Left": -2,
    "Move Right": -1,
    "Move Forward": 10,
    "Move Backward": -3,
    "Stay": 0
}

# Display action and reward mapping

print("Action-Reward Mapping")
print("-" * 35)

for action, reward in reward_values.items():
    print(action, ":", reward)
```

```python
# Environment function with five actions

def maze_environment(action):

    reward = reward_values[action]

    if action == "Move Forward":
        result = "Goal Reached!"

    elif action == "Stay":
        result = "Robot remains at the same position."

    else:
        result = "Robot moves away from the goal."

    return reward, result
```

```python
# Run five independent trials

total_reward = 0
trial_rewards = []

print("Five Trial Robot Maze Simulation")
print("=" * 45)

for trial in range(1, 6):

    # Agent selects an action
    selected_action = random.choice(actions)

    # Environment provides reward
    reward, result = maze_environment(selected_action)

    # Store reward
    trial_rewards.append(reward)

    # Calculate total reward
    total_reward += reward

    print("\nTrial:", trial)
    print("Selected Action:", selected_action)
    print("Environment Response:", result)
    print("Reward Obtained:", reward)

print("\n" + "=" * 45)
print("Individual Trial Rewards:", trial_rewards)
print("Total Reward After Five Trials:", total_reward)
```

```python
# Display trial results using Pandas

import pandas as pd

results = []

random.seed(42)

for trial in range(1, 6):

    selected_action = random.choice(actions)

    reward, result = maze_environment(selected_action)

    results.append({
        "Trial": trial,
        "Selected Action": selected_action,
        "Environment Response": result,
        "Reward": reward
    })

results_df = pd.DataFrame(results)

display(results_df)

print("Total Reward:", results_df["Reward"].sum())
```

```python
# Display trial results using Pandas

import pandas as pd

results = []

random.seed(42)

for trial in range(1, 6):

    selected_action = random.choice(actions)

    reward, result = maze_environment(selected_action)

    results.append({
        "Trial": trial,
        "Selected Action": selected_action,
        "Environment Response": result,
        "Reward": reward
    })

results_df = pd.DataFrame(results)

display(results_df)

print("Total Reward:", results_df["Reward"].sum())
```

```python
import matplotlib.pyplot as plt

# Plot rewards obtained in each trial

plt.figure(figsize=(8, 5))

plt.bar(
    results_df["Trial"],
    results_df["Reward"]
)

plt.xlabel("Trial Number")
plt.ylabel("Reward Obtained")
plt.title("Rewards Obtained in Five Robot Maze Trials")

plt.xticks(results_df["Trial"])

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Select the action with the maximum reward

best_action = max(
    reward_values,
    key=reward_values.get
)

best_reward = reward_values[best_action]

print("Reward-Based Decision Making")
print("-" * 40)

print("Available Actions:", actions)
print("Reward Values:", reward_values)

print("\nBest Action Selected:", best_action)
print("Maximum Reward:", best_reward)

print("\nResult: The robot selects Move Forward")
print("because it provides the highest reward.")
```
